```python

import string

text = "Bonjour ! J'aime le NLP, Python et l'Intelligence Artificielle... @2025"

# Convertir en minuscules
text_lower = text.lower()

# Supprimer la ponctuation
text_clean = text_lower.translate(
    str.maketrans("", "", string.punctuation)
)

print("Texte original :", text)
print("Texte nettoyé :", text_clean)
```

```python

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "Le traitement du langage naturel permet aux ordinateurs de comprendre le texte."

# Tokenisation
tokens = word_tokenize(text.lower(), language="french")
print("Tokens :", tokens)

# Charger les stopwords français
stop_words = set(stopwords.words("french"))
print("Liste des stopwords :", sorted(stop_words))

# Ajouter un mot peu utile pour notre exemple
stop_words.add("texte")

# Supprimer les stopwords et la ponctuation
filtered_tokens = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print("Mots restants :", filtered_tokens)

# Calculer la proportion supprimée
total = len(tokens)
remaining = len(filtered_tokens)
removed = total - remaining

print("Nombre total de tokens :", total)
print("Nombre de mots restants :", remaining)
print("Nombre de mots supprimés :", removed)

if total > 0:
    print("Proportion supprimée :", round(removed / total * 100, 2), "%")
```

```python

import spacy

nlp = spacy.load("fr_core_news_sm")

text = "Les étudiants étudient les différentes méthodes et les résultats."

doc = nlp(text)

print("Mots originaux :", [token.text for token in doc])
print("Lemmes :", [token.lemma_ for token in doc])
```

```python

doc = nlp(" ".join(filtered_tokens))

lemmas = [
    token.lemma_
    for token in doc
]

print("Avant lemmatisation :", filtered_tokens)
print("Après lemmatisation :", lemmas)
```
